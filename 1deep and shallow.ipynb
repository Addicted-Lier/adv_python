```python
original = [1,2,3,4]
copy = original

print(copy)
print(original)
print(id(copy))
print(id(original))

print(original is copy) # true when addr is same 
print(original == copy) # true when values are same
```

```python
# shallow copy 
import copy 
numbers = [35,45,3,456,435,346,23]

shallow = numbers.copy()

shallow.append(12)
print(shallow)
print(numbers)
```

```python
#problem with shallow copy 
#it changes the both original as well as copy
#as it copies the outer layer of the list only
import copy 
numbers = [35,45,3,[456,435,346],23]

shallow = numbers.copy()
shallow[3][1] = 12000
print(shallow)
print(numbers)
```

```python
#we use deep copy for the fix 
import copy 
numbers = [35,45,3,[456,435,346],23]

shallow = copy.deepcopy(numbers)
shallow[3][1] = 12000
print(shallow)
print(numbers)
```
