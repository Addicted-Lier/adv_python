```python
#creating a class
class Student:
    #attributes
    roll_no = 0
    name = "" 
    class_n = 0

    #methods
    # def set_details(self):
    #     self.roll_no = int(input("enter the roll no : "))
    #     self.name = input("enter your name : ")
    #     self.class_n = int(input("enter the class :"))

    #better way with parameters
    def set_details(self, roll_no:int, name:str, class_n:int) ->None:
            self.roll_no = roll_no
            self.name = name
            self.class_n = class_n

    
    def display_details(self) -> None:
        print(self.class_n)
        print(self.roll_no)
        print(self.name)


#object/instance
s1 = Student()
s1.set_details(12,"Rahul",10)
s1.display_details()

```
