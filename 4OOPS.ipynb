```python
# in this example we will remove the attributes and tryna introduce constructor
# 1. __init__ = initialize the object

# Think of __init__ as:

# "When this object is born, what information should it have?"
# now we can do Rock3 = Car("pink" , "008", "suv", "mrf")


#creating a class
class Student:
    
    #better way with parameters
    def __init__(self, roll_no:int, name:str, class_n:int) ->None:
            self.roll_no = roll_no
            self.name = name
            self.class_n = class_n

    
    def display_details(self) -> None:
        print(f"class of {self.name} is {self.class_n}")
        print(f"roll no of {self.name} is {self.roll_no}")
       


#object/instance
s1 = Student(12,"rahul", 10)

s1.display_details()

#
```

```python
#another example

class Car:
  
    #method 
    def __init__(self , color:str, model_num:str, ctype:str, tyre:str) -> None:
        self.color = color
        self.model_num = model_num
        self.ctype = ctype
        self.tyre = tyre

    def show_details(self) ->None:
        print(self.color)
        print(self.model_num)
        print(self.ctype)
        print(self.tyre)

#object or instance
rockabay=Car("Red", "007", "racing", "michilen")

rockabay.show_details()
```

```python
#student class project 

class Student:

    def __init__(self,name:str,age:int,marks:list)->None:
        self.name = name
        self.age = age
        self.marks = marks

    def std_details(self):
        print(f"Name:{self.name} Age:{self.age}")

    def average(self):
        return sum(self.marks) / len(self.marks)

    def grade(self):
        avg = self.average()
        if avg>=90 : return "A+"
        elif avg>=75 and avg<=89 : return "A"
        else : return "Fail"

s1 = Student("Anil", 17, [54,78,66,99,87])
s1.std_details()
print(s1.average())
print(s1.grade())
```
