```python
#updated version with getter and setter 

class Bank:
    def __init__(self,balance):
        self.__balance = balance
        
    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, value):
        if value > 0:
            self.__balance = value
        else:
            raise ValueError("Balance must be positive")


c1 = Bank(1200)
print(c1.balance)
c1.balance = 15000
print(c1.balance)
print(c1.__dict__)
```

```python
#lets create a real bank system where they can see balance add money 
#withdraw and deposit

class BankAccount:
    def __init__(self,amount):
        self.__amount = amount

    @property
    def amount(self):
        return self.__amount

    def deposit(self,value):
        if value>0:
            self.__amount+=value

    def withdraw(self,value):
        if value>0 and value<self.__amount:
            self.__amount-=value

c1 = BankAccount(12000)
print(c1.amount)
c1.deposit(400)
print(c1.amount)
c1.withdraw(400)
print(c1.amount)
```
